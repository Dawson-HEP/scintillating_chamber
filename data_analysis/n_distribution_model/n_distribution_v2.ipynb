# N-Distribution Model

## Imports

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

import random
```

## Create Dataframe

```python
df = pd.read_csv("data")
```

## Test

```python
start = 0
end = 160
num_repeats = 30

# Create a list with numbers from 0 to 160
numbers = list(range(start, end + 1))

# Add some random repeats
for _ in range(num_repeats):
    numbers.append(random.randint(start, end))

# Shuffle the list to mix the original numbers and the repeats
random.shuffle(numbers)

# Create a Pandas DataFrame
df = pd.DataFrame(numbers, columns=['counter'])

df
```

## Filter

```python
time = 500      #ms
clock_cycle = 20      #ms

n_data_points = time / clock_cycle


df = df.groupby("counter").head(n_data_points)

df
```

## Algorithm

```python
i = 200
j = 200


length = 40     #mm
width = 40      #mm

frequency = df["counter"].value_counts()
    
frequency

# frequency_df = pd.DataFrame({
#     "counter": frequency.index,
#     "frequency": frequency.value
# })

matrix = np.zeros((i,j))

for x in range(i - length):
    for y in range(j - width):
        matrix[x : x + length, y: y + length] += frequency[ x + y * i]
```

## Maximum

```python
maximum = matrix.max()

location = np.where(matrix == maximum)

print(f"max = {maximum}, location = {location}")
```

## Graphs

### 3D Map

```python
# Create X, Y coordinates
x = np.arange(matrix.shape[1])
y = np.arange(matrix.shape[0])
X, Y = np.meshgrid(x, y)

# Plot
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, matrix, cmap='terrain')  # 'terrain' colormap looks like mountains
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_zlabel("Frequency")
plt.show()
```

### Heat map

```python
plt.imshow(matrix, cmap='hot', interpolation='nearest')
plt.colorbar(label='Value')  # adds a color scale
plt.title("Heatmap of Matrix")
plt.xlabel("Columns")
plt.ylabel("Rows")
plt.show()
```
