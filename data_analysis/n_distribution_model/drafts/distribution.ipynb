## Rotation Matrix

```python
x = 0
y = 0
z = 0

base = np.array([x,y,z,1])


def R_z(a):
    yaw = np.array(
        [np.cos(a), -np.sin(a),     0]
        [np.sin(a), np.cos(a),      0]
        [0,         0,              1]
    )
    return yaw

def R_y(b):
    pitch = np.array(
        [np.cos(b),      0,      np.sin(b)]
        [0,              1,      0        ]
        [-np.sin(b),     0,      np.cos(b)]
    )
    return pitch

def rotate(base, a,b):

    return (base @ R_z(a)) @ R_y(b)


rotate(base,2,2)

```
