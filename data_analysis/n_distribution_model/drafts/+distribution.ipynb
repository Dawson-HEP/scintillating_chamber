## Imports

```python
import numpy as np
import matplotlib.pyplot as plt
import polars as pl
```

## Create DataFrame

```python
csv_x = [
    "sample_data.csv"
]

dfs_x = []
for i in csv_x:
    dfs_x.append(pl.read_csv(i))

```

```python
csv_y = [
    "sample_data.csv"
]

dfs = []
for i in csv_y:
    dfs.append(pl.read_csv(i))
```

## Filter

```python
rates_x = []
for df in dfs_x:
    hits = (df["trigger_data"] == 0).sum()

    rate_of_0 = hits / df.shape[0]

    rate = 1 - rate_of_0

    rates_x.append(rate)
```

```python
rates_y = []
for df in dfs_x:
    hits = (df["trigger_data"] == 0).sum()

    rate_of_0 = hits / df.shape[0]

    rate = 1 - rate_of_0

    rates_y.append(rate)
```
